```python
import pandas as pd
```

```python
t2012 = pd.read_csv("Tabela5-sem_emprego_2012.csv", sep=";", decimal=",")
t2026 = pd.read_csv("Tabela5-sem_emprego_2026.csv", sep=";", decimal=",")
```

```python
comp = t2012.merge(
    t2026,
    on=["Sigla", "Código", "Estado"],
    how="inner"
)

comp["Variação homens (p.p.)"] = (
    comp["Desocupados - homens (2026 T1)"] -
    comp["Desocupados - homens (2012 T1)"]
)

comp["Variação mulheres (p.p.)"] = (
    comp["Desocupados - mulheres (2026 T1)"] -
    comp["Desocupados - mulheres (2012 T1)"]
)

comp
```

```python
bruto = pd.read_excel("Tabela 1.1.1.xls", sheet_name="2022", header=None)

horas = bruto.iloc[8:41, [0, 1, 4, 5, 6, 7]].copy()

horas.columns = [
    "Estado",
    "Horas totais (2022)",
    "Horas homens brancos (2022)",
    "Horas homens pretos ou pardos (2022)",
    "Horas mulheres brancas (2022)",
    "Horas mulheres pretas ou pardas (2022)"
]

horas = horas[horas["Estado"].isin(comp["Estado"])].copy()

colunas_horas = horas.columns[1:]
horas[colunas_horas] = horas[colunas_horas].apply(
    pd.to_numeric,
    errors="coerce"
)

horas
```

```python
cruzado = comp.merge(horas, on="Estado", how="inner")

cruzado
```

```python
cruzado.to_csv("geral.csv", sep=";", decimal=",", index=False)

print(f"{cruzado.shape[0]} estados incluídos em geral.csv")
```
