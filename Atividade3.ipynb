## 1. Importar a biblioteca

```python
import pandas as pd
```

## 2. Ler o arquivo original

```python
arquivo = "Tabela_3_Areas_gerais_de_formacao_na_graduacao.xlsx"
bruto = pd.read_excel(arquivo, engine="openpyxl", header=None)
bruto
```

## 3. Selecionar as linhas das 5 grandes regiões

Na planilha, a linha 4 (índice 3) é o Brasil, as linhas 5 a 9 (índices 4 a 8) são as cinco grandes regiões, depois vêm os estados e por fim a linha da fonte. Vamos usar `iloc` para pegar só as linhas das regiões.

```python
dados = bruto.iloc[4:9].copy()
dados
```

## 4. Definir os nomes das colunas

```python
dados.columns = [
    "regiao",
    "total_superior", "homens_superior", "mulheres_superior",
    "total_ctem", "homens_ctem", "mulheres_ctem",
    "total_bem_estar", "homens_bem_estar", "mulheres_bem_estar",
]
dados
```

## 5. Converter as colunas numéricas

```python
colunas_numericas = [
    "total_superior", "homens_superior", "mulheres_superior",
    "total_ctem", "homens_ctem", "mulheres_ctem",
    "total_bem_estar", "homens_bem_estar", "mulheres_bem_estar",
]

for coluna in colunas_numericas:
    dados[coluna] = pd.to_numeric(dados[coluna], errors="coerce")

dados
```

## 6. Confirmar que só temos as 5 grandes regiões

Usamos `isin()` para garantir que ficaram apenas Norte, Nordeste, Sudeste, Sul e Centro-Oeste (sem Brasil, sem estados e sem a linha da fonte).

```python
regioes = ["Norte", "Nordeste", "Sudeste", "Sul", "Centro-Oeste"]
dados = dados[dados["regiao"].isin(regioes)]
dados
```

## 7. Tabela original já tratada

Esta é a tabela `dados`, já com as colunas nomeadas, os valores numéricos convertidos e apenas as 5 grandes regiões.

```python
dados
```

## 8. Tabela 1 — Homens e mulheres com curso superior completo

```python
tabela_superior = dados[["regiao", "homens_superior", "mulheres_superior"]].copy()
tabela_superior = tabela_superior.rename(columns={
    "regiao": "Regiao",
    "homens_superior": "Homens",
    "mulheres_superior": "Mulheres",
})
tabela_superior = tabela_superior.reset_index(drop=True)
tabela_superior
```

```python
tabela_superior.describe()
```

## 9. Tabela 2 — Homens e mulheres em Ciência, Tecnologia, Engenharias e Matemática

```python
tabela_ctem = dados[["regiao", "homens_ctem", "mulheres_ctem"]].copy()
tabela_ctem = tabela_ctem.rename(columns={
    "regiao": "Regiao",
    "homens_ctem": "Homens",
    "mulheres_ctem": "Mulheres",
})
tabela_ctem = tabela_ctem.reset_index(drop=True)
tabela_ctem
```

```python
tabela_ctem.describe()
```

## 10. Tabela 3 — Homens e mulheres em Educação, Serviços pessoais, Saúde e Bem-estar

```python
tabela_bem_estar = dados[["regiao", "homens_bem_estar", "mulheres_bem_estar"]].copy()
tabela_bem_estar = tabela_bem_estar.rename(columns={
    "regiao": "Regiao",
    "homens_bem_estar": "Homens",
    "mulheres_bem_estar": "Mulheres",
})
tabela_bem_estar = tabela_bem_estar.reset_index(drop=True)
tabela_bem_estar
```

```python
tabela_bem_estar.describe()
```

## 11. Salvar as três tabelas em arquivos .xlsx

```python
tabela_superior.to_excel("superior_por_regiao_julio.xlsx", index=False)
tabela_ctem.to_excel("ctem_por_regiao_julio.xlsx", index=False)
tabela_bem_estar.to_excel("educacao_saude_bem_estar_julio.xlsx", index=False)

print("Os três arquivos foram salvos com sucesso:")
print("- superior_por_regiao_julio.xlsx")
print("- ctem_por_regiao_julio.xlsx")
print("- educacao_saude_bem_estar_julio.xlsx")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
brasil_dados = bruto.iloc[[3]].copy()

brasil_dados.columns = [
    "regiao",
    "total_superior", "homens_superior", "mulheres_superior",
    "total_ctem", "homens_ctem", "mulheres_ctem",
    "total_bem_estar", "homens_bem_estar", "mulheres_bem_estar",
]

for coluna in brasil_dados.columns[1:]:
    brasil_dados[coluna] = pd.to_numeric(brasil_dados[coluna], errors="coerce")

brasil = pd.DataFrame({
    "area": ["Graduação (total)", "STEM", "Saúde / educação"],
    "homens": [
        brasil_dados["homens_superior"].iloc[0],
        brasil_dados["homens_ctem"].iloc[0],
        brasil_dados["homens_bem_estar"].iloc[0]
    ],
    "mulheres": [
        brasil_dados["mulheres_superior"].iloc[0],
        brasil_dados["mulheres_ctem"].iloc[0],
        brasil_dados["mulheres_bem_estar"].iloc[0]
    ]
})

brasil_long = brasil.melt(
    id_vars="area",
    var_name="sexo",
    value_name="pessoas"
)

fig, ax = plt.subplots(figsize=(9, 5))
sns.barplot(data=brasil_long, x="area", y="pessoas", hue="sexo", ax=ax)
ax.set_title("Brasil: pessoas com graduação, por área e sexo")
ax.set_xlabel("Área")
ax.set_ylabel("Pessoas")
plt.tight_layout()
plt.show()
```
